```python
from pathlib import Path
import pandas as pd

results = Path("results/paper")
accuracy = pd.read_csv(results / "table_iv_accuracy.csv")
mia = pd.read_csv(results / "table_v_mia_max_softmax.csv")
accuracy.head()
```

```python
operating_points = accuracy.merge(mia, on=["dataset", "backbone", "configuration"])
operating_points.pivot_table(index=["dataset", "backbone"], columns="configuration", values="test_accuracy_percent")
```

```python
import matplotlib.pyplot as plt

for setting, group in operating_points.groupby("configuration"):
    plt.scatter(group["oriented_auc"], group["test_accuracy_percent"], label=setting)
plt.axvline(0.5, color="grey", linestyle="--", linewidth=1)
plt.xlabel("Oriented membership ROC-AUC")
plt.ylabel("Test accuracy (%)")
plt.legend()
plt.show()
```
